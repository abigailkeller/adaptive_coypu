```python
# import libraries
from dd.autoref import BDD
from itertools import product
import math
```

```python
# create states and actions
N_STATES   = 4
STATE_VALS = [0, 50, 100, 150]
VARS       = ["X", "Y", "Z"]
ACTION_COST = {"a1": 10, "a2": 10, "a3": 0}
```

```python
# build ADD class

from __future__ import annotations
from dataclasses import dataclass
from typing import Any

# ── nodes ──────────────────────────────────────────────────────────────────────

@dataclass(frozen=True)
class Leaf:
    value: float

    def __repr__(self):
        return f"Leaf({self.value})"

@dataclass(frozen=True)
class Node:
    var:      str
    children: tuple  # one per state value, ordered by STATE_VALS

    def __repr__(self):
        return f"Node({self.var})"

# ── manager (handles hash-consing) ────────────────────────────────────────────

class ADD:
    def __init__(self, state_vals: list):
        self.state_vals  = state_vals
        self._leaf_cache = {}   # value -> Leaf
        self._node_cache = {}   # (var, children_tuple) -> Node

    def leaf(self, value: float) -> Leaf:
        if value not in self._leaf_cache:
            self._leaf_cache[value] = Leaf(value)
        return self._leaf_cache[value]

    def node(self, var: str, children: tuple) -> Leaf | Node:
        # Reduction rule: all children identical -> return child
        if len(set(id(c) for c in children)) == 1:
            return children[0]
        key = (var, children)
        if key not in self._node_cache:
            self._node_cache[key] = Node(var, children)
        return self._node_cache[key]

    def build(self, fn, var_order: list) -> Leaf | Node:
        """Build an ADD from a function f(assignment) -> float."""
        def _build(depth, assignment):
            if depth == len(var_order):
                return self.leaf(fn(assignment))
            var = var_order[depth]
            children = tuple(
                _build(depth + 1, {**assignment, var: v})
                for v in self.state_vals
            )
            return self.node(var, children)
        return _build(0, {})

    def evaluate(self, root: Leaf | Node, assignment: dict) -> float:
        """Evaluate ADD at a given assignment."""
        node = root
        while isinstance(node, Node):
            val = assignment[node.var]
            idx = self.state_vals.index(val)
            node = node.children[idx]
        return node.value

    def apply(self, op, a: Leaf | Node, b: Leaf | Node) -> Leaf | Node:
        """Apply a binary operation to two ADDs."""
        cache = {}
        def _apply(x, y):
            key = (id(x), id(y))
            if key in cache:
                return cache[key]
            if isinstance(x, Leaf) and isinstance(y, Leaf):
                result = self.leaf(op(x.value, y.value))
            elif isinstance(x, Node) and isinstance(y, Node) and x.var == y.var:
                result = self.node(x.var, tuple(
                    _apply(cx, cy)
                    for cx, cy in zip(x.children, y.children)
                ))
            elif isinstance(x, Node):
                result = self.node(x.var, tuple(
                    _apply(cx, y) for cx in x.children
                ))
            else:
                result = self.node(y.var, tuple(
                    _apply(x, cy) for cy in y.children
                ))
            cache[key] = result
            return result
        return _apply(a, b)

    def count_nodes(self, root: Leaf | Node) -> int:
        seen = set()
        def _walk(n):
            if id(n) in seen:
                return 0
            seen.add(id(n))
            if isinstance(n, Leaf):
                return 1
            return 1 + sum(_walk(c) for c in n.children)
        return _walk(root)

    def count_leaves(self, root: Leaf | Node) -> int:
        seen = set()
        def _walk(n):
            if id(n) in seen:
                return 0
            seen.add(id(n))
            if isinstance(n, Leaf):
                return 1
            return sum(_walk(c) for c in n.children)
        return _walk(root)

    def to_dot(self, root: Leaf | Node, title: str = "ADD") -> str:
        """Build an ADD from a function f(assignment) -> float."""
    lines = [
        f'digraph "{title}" {{',
        '  node [fontname="Helvetica"];',
        '  edge [fontname="Helvetica"];',
    ]
    seen = set()

        def node_id(n):
            return f"n{id(n)}"

        def _walk(n):
            if id(n) in seen:
                return
            seen.add(id(n))
            nid = node_id(n)
            if isinstance(n, Leaf):
                lines.append(
                    f'  {nid} [shape=square, style=filled, '
                    f'fillcolor=lightblue, label="{n.value}"];'
                )
            else:
                lines.append(f'  {nid} [shape=ellipse, label="{n.var}"];')
                for val, child in zip(self.state_vals, n.children):
                    lines.append(
                        f'  {nid} -> {node_id(child)} [label="{val}"];'
                    )
                    _walk(child)

        _walk(root)
        lines.append("}")
        return "\n".join(lines)
```

```python
# reward function
def reward_fn(s, a):
    return s["X"] + s["Y"] + s["Z"] + ACTION_COST[a]
```

```python
import operator

# Build
mgr = ADD(STATE_VALS)

reward_adds = {
    a: mgr.build(lambda s, a=a: reward_fn(s, a), VARS)
    for a in ACTION_COST
}
```

```python
# evaluate
print(mgr.evaluate(reward_adds["a1"], {"X": 50, "Y": 100, "Z": 150}))
```

```python
# Node counts
for a, add in reward_adds.items():
    print(f"{a}: {mgr.count_nodes(add)} nodes, {mgr.count_leaves(add)} leaves")
```

4 possible state values, each variable needs to be encoded in binary to work with the BDD library, which only understands boolean (true/false) variables.

STATE_VALS[0] =   0  →  index 0  →  bits: x0=False, x1=False  (00)

STATE_VALS[1] =  50  →  index 1  →  bits: x0=True,  x1=False  (01)

STATE_VALS[2] = 100  →  index 2  →  bits: x0=False, x1=True   (10)

STATE_VALS[3] = 150  →  index 3  →  bits: x0=True,  x1=True   (11)

```python
# ── encode multi-valued variables as binary ────────────────────────────────────
# 4 states needs 2 bits per variable
# X -> x0, x1  (x0 is LSB)
# Y -> y0, y1
# Z -> z0, z1

N_BITS = int(math.log2(N_STATES))

def var_bits(var):
    """Return bit variable names for a state variable, LSB first."""
    return [f"{var.lower()}{i}" for i in range(N_BITS)]

all_bit_vars = [b for var in VARS for b in var_bits(var)]

# Map state value -> bit assignment {bit_var: bool}
def val_to_bits(var, val):
    idx = STATE_VALS.index(val)
    bits = var_bits(var)
    return {bits[i]: bool((idx >> i) & 1) for i in range(N_BITS)}

# Map bit assignment back to state value
def bits_to_val(var, assignment):
    bits = var_bits(var)
    idx = sum(int(assignment[b]) << i for i, b in enumerate(bits))
    return STATE_VALS[idx]
```

```python
def build_reward_add(bdd, action):
    """
    Build reward ADD for a given action by summing over all assignments.
    Each terminal value is encoded as a constant ADD node.
    We build it as a sum of indicator functions:
        ADD = sum over all (x,y,z): reward(x,y,z,a) * [[X=x, Y=y, Z=z]]
    where [[...]] is the indicator BDD for that assignment.
    """
    result = bdd.add_expr("False")  # start with 0 (additive identity)

    for x, y, z in product(STATE_VALS, repeat=3):
        r = reward_fn(x, y, z, action)

        # Build indicator BDD for this exact assignment
        assignment = {}
        assignment.update(val_to_bits("X", x))
        assignment.update(val_to_bits("Y", y))
        assignment.update(val_to_bits("Z", z))

        # Conjoin all bit literals to get indicator for this state
        literals = [
            f"{'' if v else '~'}{bit}"
            for bit, v in assignment.items()
        ]
        indicator = bdd.add_expr(" & ".join(literals))

        # ADD: result += r * indicator
        # In dd, we use bdd.apply for ADD arithmetic
        const_r  = bdd.add_expr(str(r))   # constant ADD node
        term     = bdd.apply("*", const_r, indicator)
        result   = bdd.apply("+", result, term)

    return result
```

```python
# ── run it ─────────────────────────────────────────────────────────────────────
bdd = BDD()
bdd.declare(*all_bit_vars)

reward_adds = {}
for action in ACTION_COST:
    reward_adds[action] = build_reward_add(bdd, action)
    print(f"Action {action}: {len(reward_adds[action])} nodes")

```

```python
# ── evaluate at a specific state ───────────────────────────────────────────────
def evaluate(add, bdd, x, y, z):
    assignment = {}
    assignment.update(val_to_bits("X", x))
    assignment.update(val_to_bits("Y", y))
    assignment.update(val_to_bits("Z", z))
    return bdd.let(assignment, add)

result = evaluate(reward_adds["a1"], bdd, x=50, y=100, z=150)
```
